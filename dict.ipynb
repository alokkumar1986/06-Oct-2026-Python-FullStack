```python
cars = { "car1": {
        "make": "Toyota",
        "model": "Camry",
        "year": 2024,
        "features": ["Bluetooth", "Backup Camera", "Lane Assist", "Heated Seats"],
        "dimensions_inches": (192.1, 72.4, 56.9),
        "engine": {
            "type": "Hybrid",
            "horsepower": 208,
            "transmission": "eCVT"
        }
    },
    "car2" : {
        "make": "Honda",
        "model": "Civic",
        "year": 2023,
        "features": ["Apple CarPlay", "Sunroof", "Adaptive Cruise Control"],
        "dimensions_inches": (184.0, 70.9, 55.7),
        "engine": {
            "type": "Gasoline",
            "horsepower": 158,
            "transmission": "CVT"
        }
    },
    "car3" : {
        "make": "Ford",
        "model": "Mustang",
        "year": 2024,
        "features": ["Track Apps", "Leather Seats", "Premium Audio"],
        "dimensions_inches": (189.4, 75.4, 55.0),
        "engine": {
            "type": "V8",
            "horsepower": 480,
            "transmission": "10-Speed Automatic"
        }
    }
}

print(cars['car3']['engine']['horsepower'])

print(cars['car1']['dimensions_inches'])
print(cars['car2']['dimensions_inches'])
print(cars['car3']['dimensions_inches'])

for i in cars.values():
  print(i['dimensions_inches'])

for (i,j) in cars.items():
  print(j['dimensions_inches'])
```
