```python
#shallow copy
A = {1, 2, 3, 4, 5}

B = A.copy()
print(B)

B.add(6)
print(B)

print(A)
```

```python
#deep copy
A = {1, 2, 3, 4, 5}

B = A
print(B)

B.add(6)
print(B)

print(A)
```
