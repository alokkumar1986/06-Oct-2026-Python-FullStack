```python
t = (1, 2, 3, 4, 5)

print(t[0]) #1
print(t[1]) #2
print(t[2]) #3
print(t[3]) #4
print(t[4]) #5
#print(t[5]) #IndexError

print(t[-1]) #5
print(t[-2]) #4
print(t[-3]) #3
print(t[-4]) #2
print(t[-5]) #1
print(t[-6]) #IndexError

```

```python
t = (1, 2, 3, 4, 5, 6, 7, 8, 9)

print(t[:])

print(t[2:])

print(t[:5])

print(t[2:6])

print(t[::-1])

print(t[2:100])

print(t[10:2]) #() empty tuple

print(t[-10:-1]) #

print(t[:-1])


```
