```python
#arithmatic operations.

a = 10
b = 15

print(a+b)
print(a-b)
print(a*b)
print(a/b)
print(a%b)
print(a**b)
print(a//b)
```

```python
#comparison operator

a = 10
b = 15
print(a>b)
print(a<b)
print(a>=b)
print(a<=b)
print(a==b)
print(a!=b)
```

```python
print(10 < 15 < 5) #chaining of comparison operator
```

```python
print(10 > 15 < 5 == 6)
```

```python
#assignment operators

a = 15
b = 6

a +=1    # a = a+1
print(a)  #16

a-=1   # a = a-1
print(a) #15

a*=b  # a = a*b
print(a) #9

a/=b   #a = a/b
print(a) #15.0

a%=b   # a = a%b
print(a) #3.0

a**=b # a= a**b
print(a) #729.0

a//=b  # a = a//b
print(a)
```

```python
a = 10
b = 0

print(a and b)
print(b and a)

print(a or b)
print(b or a)

print(not a)
print(not b)
```

```python

#membership operator

print(5 in range(10))
print(5 in [1, 2, 3, 4])
print("A" not in "Aptech")


for i in range(10):
  print(i)
```
