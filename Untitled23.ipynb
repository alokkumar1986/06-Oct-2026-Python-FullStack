# ARITHMETIC OPERATOR

```python
a=10
b=15
print(a+b)
print(a-b)
print(a*b)
print(a/b)
print(a%b)
print(a**b)
print(a//b)
```

LOGICAL OPERATOR

```python
a = 10
b = 0
print(a and b)
print(b and a)
print(a or b)
print(b or a)

print(not a)
print(not b)
```

MEMBERSHIP OPERATOR

```python
print(5 in range (10))
print( 5 in [1,2,3,4])
print("A" not in "Aptech")


for i in range (10) :
  print(i)
```
