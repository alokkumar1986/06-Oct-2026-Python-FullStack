# ***slicing***

```python
t = (1,2,3,4,5,6,7,8,9)
print(t[:])
print(t[2:])
print(t[:5])
print(t[2:6])
print(t[::-1])
print(t[2:100])
print(t[10:2])
print(t[-10:-1])
print(t[:-1])
print(t[-8:-2:-1])
print(t[-2:-8:-1])
print(t[8:2:-1])
print(t[2:8:-1])
print(t[8:2])
```

```python
print(t[-8:-2:-1])
print(t[-2:-8:-1])
print(t[8:2:-1])
print(t[2:8:-1])
print(t[8:2])
print(t[2:8])
```
