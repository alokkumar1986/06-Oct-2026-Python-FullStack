Take input from keyboard to find the numbers which are divisible by 3 between any two numbers

```python
num1 = int(input('enter num1 :'))
num2 =int(input('enter num2 :'))
for i in range (num1,num2+1):
  if i%3 ==0:
    print(i,end=' ')
```
