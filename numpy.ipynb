```python
#array()
import numpy as np
x = np.array([1, 2, 3, 4])
print(type(x))
print(x.shape)
```

```python
#arange()

import numpy as np

x = np.arange(1, 100, 5)
print(x)
```

```python
#linspace()

import numpy as np
x = np.linspace(1, 10, 5)
print(x)
```

```python
#empty()

import numpy as np

x = np.empty(4)

print(x)
```

```python
#zeros()
#ones()

import numpy as np

x= np.zeros((2,2))
print(x)

y = np.ones((2,2))
print(y)
```

```python
import numpy as np

x = y = np.array([
    [
        [1, 2, 3, 4],
        [2, 3, 4, 5]
    ],
    [
        [3, 4, 5, 6],
        [4, 5, 6, 7]
    ]
    ])

print(x[0:1, 1:, 3:])

print(x[1:, 0:1, 1:3])
```

```python
import numpy as np

x = np.array([1, 4, 9, 3])
y = np.array([6, 5, 7, 2])

z = np.hstack((x, y))
print(z)

p = np.vstack((x, y))
print(p)

q = np.stack((x, y), axis=1)  #axis = 0 ,x, y (horizontal, then join with vstack())
                              #axis = 1, x, y (vertical and join with hstack())
print(q)

r = np.concatenate((x, y))
print(r)
```

```python
import numpy as np

arr = np.arange(24).reshape(4, 6)

print(arr)

print(np.hsplit(arr, 6))


```

```python
import numpy as np

arr = np.arange(24).reshape(4, 6)

print(arr)

print(np.vsplit(arr, 4))
```

```python
import numpy as np

arr = np.arange(24).reshape(4, 6)

print(arr)

print(np.split(arr, 6, axis=1))
```

```python
import numpy as np

arr = np.arange(24).reshape(4, 6)

print(arr+1)

print(np.add(arr, 1))

print(arr-1)

print(np.subtract(arr, 1))

print(arr*2)

print(np.multiply(arr, 2))

print(arr/2)

print(np.divide(arr, 2))

print(arr%2)

print(np.mod(arr, 2))
```
